```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
print(tf.__version__)
```

```python
trainig_data = np.load('TF_dataset.npz')
input_size = 2
output_size = 1
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(5, activation="sigmoid"),
    tf.keras.layers.Dense(5, activation="sigmoid"),
    tf.keras.layers.Dense(output_size)
])
model.compile(optimizer='adam', loss='mean_absolute_error')
model.fit(trainig_data['inputs'], trainig_data['targets'], epochs=100, verbose=1)
```

```python
model.layers[0].get_weights()
```

```python
plt.plot(np.squeeze(model.predict_on_batch(trainig_data['inputs'])),
        np.squeeze(trainig_data['targets']), color='orange')
plt.xlabel('outputs', fontsize=10)
plt.ylabel('targets', fontsize=10)
plt.show()
```
